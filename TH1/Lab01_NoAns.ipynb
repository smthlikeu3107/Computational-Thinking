### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
def area(a, b, c):
    a, b, c = sorted([a, b, c])
    if a + b <= c or a <= 0:
        raise ValueError("Invalid triangle sides")
    p = (a + b + c) / 2
    return (p * (p - a) * (p - b) * (p - c))**0.5

print(area(3, 4, 5))
print(area(5, 12, 13))
print(area(1, 2, 1))
```

### Problem 7
#### Write a function that returns True if a, b, c can be the side lengths of a triangle; otherwise return False.

```python
def is_triangle(a, b, c):
    a, b, c = sorted([a, b, c])
    return a + b > c and a > 0
```

### Problem 8
#### Write a function that returns the character that is n positions after character x in the alphabet.
#### Preserve case and wrap around within A..Z or a..z.

```python
def nth_char(ch, n):
    return chr(ord(ch) + n)

print(nth_char('K', 1))
```

```python
def nth_char(ch: str, n: int) -> str:
    if (not isinstance(ch, str) or len(ch) != 1):
        raise ValueError("Must be a single character")
    
    isupper = False
    if (ch.isupper()):
        isupper = True
        ch = ch.lower()
    ch = chr((ord(ch) - ord('a') + n) % 26 + ord('a'))
    if (isupper):
        ch = ch.upper()
    return ch



print(nth_char('a', 28))
```

### Problem 9
#### Write a function that converts an uppercase letter to lowercase. If the input is not a single character, raise ValueError.

```python
def tolower(ch: str) -> str:
    if (not isinstance(ch, str) or len(ch) != 1):
        raise ValueError("Must be a single character")
    return ch.lower()


print(tolower('1'))
```

### Problem 12
#### Write a function that returns a right isosceles triangle of height n using '*'.
#### Line i (1-indexed) has i '*', joined with newlines.
#### Example with n = 4
#### *
#### **
#### ***
#### ****

```python
n = int(input())
for i in range(1, n + 1):
    print('*' * i)
```
