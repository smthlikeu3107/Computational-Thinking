```python
#Bai 3
c = input()
if (ord('a') <= ord(c) <= ord('z')):
    c = chr(ord(c) - ord('a') + ord('A'))
else:
    c = chr(ord(c) - ord('A') + ord('a'))
print(c)
```

```python
#Bai 4
c = input()
if (len(c) == 1 and (ord('a') <= ord(c) <= ord('z') or ord('A') <= ord(c) <= ord('Z'))):
    print(f'{c} la ki tu alphabet')
else:
    print(f'{c} khong phai la ki tu alphabet')
```
