```python
#Bai 6
a, b, c = sorted(map(int, input().split()))
if (a + b > c):
    p = (a + b + c) / 2
    s = (p * (p - a) * (p - b) * (p - c)) ** 0.5
    print(s)
else:
    print('Khong phai 3 canh cua tam giac')
```

```python
#Bai 8
import math

ex_rate = {
    (0, 50): 1984,
    (50, 100): 2050,
    (100, 200): 2380,
    (200, 300): 2998,
    (300, 400): 3350,
    (400, math.inf): 3460,
}

def calculate_cost(kw: int) -> int:
    cost = 0
    for (lower, upper), rate in ex_rate.items():
        if lower > kw:
            break
        cost += (min(kw, upper) - lower) * rate
    return int(cost * 1.08)


name = input("Ten chu ho: ")
val1 = int(input("Chi so thang truoc: "))
val2 = int(input("Chi so thang nay: "))

print(f'Ho va ten: {name}')
print(f'Tien phai tra la: {calculate_cost(val2 - val1)}')
```
