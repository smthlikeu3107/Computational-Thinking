```python
#BAI 3
n = int(input())
for i in range(1, 10):
    print(f"{n} x {i} = {n * i}")
```

```python
h = int(input())
print(' ' * (h - 1) + '*')
for i in range(2, h):
    print(' ' * (h - i) + '*' + ' ' * (2 * (i - 1) - 1) + '*')
print('*' * (2 * h - 1))
```

```python
n = int(input())
print(n * (n + 1) // 2)
```

```python
#BAI 8
a, b = map(int, input().split())
a, b = sorted((a, b))
for i in range(2, a + 1):
    if a % i == 0 and b % i == 0:
        print(i)
```

```python
#BAI 9
n = int(input())
root = int(n ** 0.5)
print("YES" if root * root == n else "NO")
```

```python
#BAI 10
n = int(input())
sum = 0
for i in range(1, n // 2 + 1):
    if n % i == 0:
        sum += i
print("YES" if sum == n else "NO")
```
