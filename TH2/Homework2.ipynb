```python
#Bai 5
a, b, c = map(int, input().split())

def ptb1(a, b):
    if a == 0:
        if b == 0:
            return "Phuong trinh vo so nghiem"
        else:
            return "Phuong trinh vo nghiem"
    else:
        x = -b / a
        return f"Phuong trinh co nghiem duy nhat: x = {x}"

if a == 0:
    print(ptb1(b, c))
else:
    delta = b * b - 4 * a * c
    if delta < 0:
        print("Phuong trinh vo nghiem")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"Phuong trinh co nghiem kep: x1 = x2 = {x}")
    else:
        x1 = (-b + delta ** 0.5) / (2 * a)
        x2 = (-b - delta ** 0.5) / (2 * a)
        print(f"Phuong trinh co 2 nghiem phan biet: x1 = {x1}, x2 = {x2}")
```

```python
#Bai 8
a, b, m = map(int, input().split())
c, d, n = map(int, input().split())

D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c

if D != 0:
  x = Dx / D
  y = Dy / D
#   x = 0.0 if x == 0 else x
#   y = 0.0 if y == 0 else y
  print(f"Hệ phương trình có nghiệm duy nhất: x = {x:.2f}, y = {y:.2f}")
else:
  if (a == 0 and b == 0 and m != 0) or (c == 0 and d == 0 and n != 0):
    print("Hệ phương trình vô nghiệm.")
  elif Dx == 0 and Dy == 0:
    print("Hệ phương trình có vô số nghiệm.")
  else:
    print("Hệ phương trình vô nghiệm.")
```

```python
#Bai 10
x, y, r = map(int, input().split())
x1, y1 = map(int, input().split())
if (x1 - x) ** 2 + (y1 - y) ** 2 == r ** 2:
    print("Diem nam tren duong tron")
elif (x1 - x) ** 2 + (y1 - y) ** 2 < r ** 2:
    print("Diem nam trong duong tron")
else:
    print("Diem nam ngoai duong tron")
```

```python
#Bai 11
x, y = map(int, input().split())
def power(x, y) -> int:
    if y == 0:
        return 1
    if y % 2 == 0:
        return power(x, y // 2) * power(x, y // 2)
    else:
        return x * power(x, y // 2) * power(x, y // 2)

ans1 = x ** y
ans2 = power(x, y)
print(ans1)
print(ans2)
```
